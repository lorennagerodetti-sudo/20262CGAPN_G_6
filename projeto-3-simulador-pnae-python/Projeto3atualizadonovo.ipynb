```python
# ===== SIMULADOR PNAE - PROJETO 3 =====

# Valores por aluno por dia letivo (R$)
# Fonte: Resolução CD/FNDE nº 1, de 18/02/2026
# Publicação: DOU de 19/02/2026

valorCreche = 1.57
valorPreEscola = 0.82
valorFundamental = 0.57


# Custo dos alimentos por refeição (R$)
# Valor: R$ 1,44
# Fonte: Brandão, I. F. (2020).
# Modelo conceitual de sistema de gestão de custos
# da alimentação escolar no setor público.
# XXVII Congresso Brasileiro de Custos.
# Dado referente ao campus Baturité do IFCE, maio de 2018.
#
# Justificativa:
# Não existe valor oficial nacional para o custo exclusivo
# dos alimentos por refeição.
#
# Limitações:
# O valor é de um campus federal, não de uma escola
# municipal de São Paulo, e utiliza dados de 2018.

custoRefeicao = 1.44


# ===== FUNÇÕES =====
def validaNumero(mensagem):
    numero = input(mensagem)
    while not numero.isdigit():
        print("Você digitou número incorreto")
        numero = input(mensagem)
    return int(numero)


def calculaRepasse(creche, preEscola, fundamental, diasLetivos):
    valorDia = creche * valorCreche + preEscola * valorPreEscola + fundamental * valorFundamental
    repasse = valorDia * diasLetivos
    return repasse


def classifica_porte(totalMatriculas):
    if totalMatriculas <= 200:
        porte = "Pequena"
    elif totalMatriculas <= 400:
        porte = "Média"
    else:
        porte = "Grande"
    return porte


def calculaCusto(totalMatriculas, diasLetivos, custoRefeicao):
    custo = totalMatriculas * diasLetivos * custoRefeicao
    return custo


def calculaCobertura(repasse, custo):
    if custo == 0:
        cobertura = 0
    else:
        cobertura = repasse / custo * 100
    return cobertura


# ===== PROGRAMA PRINCIPAL =====
print("Olá, bem vinda ao simulador PNAE***")
print()
print("***Dados da escola***")
nome = input("Digite seu nome: ")
escola = input("Digite o nome da escola: ")

diasLetivos = validaNumero("Quantidade de dias letivos no ano: ")
creche = validaNumero("Matrículas na creche: ")
preEscola = validaNumero("Matrículas na pré-escola: ")
fundamental = validaNumero("Matrículas no ensino fundamental: ")

totalMatriculas = creche + preEscola + fundamental
porte = classifica_porte(totalMatriculas)
repasse = calculaRepasse(creche, preEscola, fundamental, diasLetivos)
custo = calculaCusto(totalMatriculas, diasLetivos, custoRefeicao)
cobertura = calculaCobertura(repasse, custo)

print()
print("Responsável:", nome)
print("Escola:", escola)
print("Total de matrículas:", totalMatriculas)
print("Porte da escola:", porte)
print("Repasse anual do PNAE: R$", round(repasse, 2))
print("Custo anual estimado: R$", round(custo, 2))
print("Cobertura do repasse:", round(cobertura, 1), "%")

if repasse >= custo:
    print("O repasse cobre o custo. Diferença a cobrir: R$ 0")
else:
    diferenca = custo - repasse
    print("Diferença a cobrir: R$", round(diferenca, 2))
```
