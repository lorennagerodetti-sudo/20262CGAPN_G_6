```python
### =====================================================================
# Simulador PNAE – Projeto 3
# =====================================================================

# ===== PARÂMETROS =====
# Fonte dos valores per capita (R$ por aluno por dia letivo):
# Resolução CD/FNDE nº 1, de 18/02/2026 (planilha do Projeto 1).
VALOR_CRECHE = 1.57
VALOR_PRE_ESCOLA = 0.82
VALOR_FUNDAMENTAL = 0.57

# Faixas de porte (mínimo de matrículas, porte): classificação didática do curso.
FAIXAS_PORTE = [(0, "Pequena"), (201, "Média"), (401, "Grande")]

# Custo só dos ALIMENTOS por refeição, em R$ (use ponto decimal).
# Fonte: Brandão, I. F. "Modelo conceitual de sistema de gestão de custos da
# alimentação escolar no setor público". XXVII Congresso Brasileiro de Custos,
# 2020. https://anaiscbc.emnuvens.com.br/anais/article/download/4758/4779/4896
# Valor: custo padrão de gêneros alimentícios por refeição (R$ 1,44), campus
# Baturité do IFCE, maio/2018. Não existe valor oficial nacional (o FNDE
# informou em 2025 que ainda não tem esse dado); este é um valor de referência.
CUSTO_ALIMENTOS_REFEICAO = 1.44


# ===== FUNÇÕES =====
#### Programa Principal
print("Olá, bem vinda ao simulador PNAE***")
print()
print ("***Dados pessoais***")
nome = input("Digite seu nome: ")
escola = input("Digite o nome da escola: ")

def ler_numero(mensagem):
    """Pede um número inteiro >= 0 e repete até o usuário digitar certo."""
    texto = input(mensagem).strip()
    while not texto.isdigit():
        print("Você digitou um número incorreto. Use apenas dígitos, sem sinal.")
        texto = input(mensagem).strip()
    return int(texto)


def calcular_repasse(creche, pre_escola, fundamental, dias_letivos):
    """Repasse anual = soma(matrículas x valor per capita) x dias letivos."""
    valor_por_dia = (creche * VALOR_CRECHE
                     + pre_escola * VALOR_PRE_ESCOLA
                     + fundamental * VALOR_FUNDAMENTAL)
    return valor_por_dia * dias_letivos


def classificar_porte(total_matriculas):
    """Devolve o porte conforme as faixas (busca aproximada, como o PROCV)."""
    porte = FAIXAS_PORTE[0][1]
    for minimo, nome in FAIXAS_PORTE:
        if total_matriculas >= minimo:
            porte = nome
    return porte


def calcular_custo_anual(total_alunos, dias_letivos, custo_refeicao):
    """Custo anual = alunos x dias letivos x custo por refeição
    (1 refeição por aluno por dia)."""
    return total_alunos * dias_letivos * custo_refeicao


def calcular_cobertura(repasse, custo):
    """Devolve (cobertura em %, diferença a cobrir, repasse cobre o custo?)."""
    if custo <= 0:
        return 0.0, 0.0, True  # sem custo, não há o que cobrir
    cobertura = repasse / custo * 100
    if repasse >= custo:
        return cobertura, 0.0, True
    return cobertura, custo - repasse, False


def formatar_reais(valor):
    """1234.5 -> 'R$ 1.234,50'"""
    texto = f"{valor:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")
    return "R$ " + texto


# ===== PROGRAMA PRINCIPAL =====
def main():
    print("=== Simulador PNAE ===")
    dias = ler_numero("Dias letivos no ano: ")
    creche = ler_numero("Matrículas na creche: ")
    pre = ler_numero("Matrículas na pré-escola: ")
    fundamental = ler_numero("Matrículas no ensino fundamental: ")

    total = creche + pre + fundamental
    repasse = calcular_repasse(creche, pre, fundamental, dias)
    porte = classificar_porte(total)
    custo = calcular_custo_anual(total, dias, CUSTO_ALIMENTOS_REFEICAO)
    cobertura, diferenca, cobre = calcular_cobertura(repasse, custo)

    print()
    print("Total de matrículas:      ", total)
    print("Porte da escola:          ", porte)
    print("Repasse anual do PNAE:    ", formatar_reais(repasse))
    print("Custo anual estimado:     ", formatar_reais(custo))
    print("Cobertura do repasse:      " + f"{cobertura:.1f}%".replace(".", ","))
    if cobre:
        print("O repasse cobre o custo. Diferença a cobrir: R$ 0,00")
    else:
        print("Diferença a cobrir:       ", formatar_reais(diferenca))


main()
```
